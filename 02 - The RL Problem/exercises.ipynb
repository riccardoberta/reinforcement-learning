## Exercises on the Reinforcement Learning Problem

These activities reinforce the core ideas presented in the RL problem chapter: the MDP tuple, the transition and reward functions and their normalisation, the return, discounting and its recursive form, and the notion of a policy.

### Exercise 2.1 — Building a transition distribution (slippery grid)

Consider the FrozenLake grid world with cells numbered. Movement is slippery: the intended direction happens with probability 1/3, and each of the two orthogonal directions also with probability 1/3. A move into a wall keeps the agent in place. For the starting state and the action LEFT, write out the full transition distribution and verify it is a valid probability distribution.

**Step 1 — Enumerate the three equally likely slip directions** 

For action LEFT, the intended direction is LEFT; the two orthogonal directions are UP and DOWN, each with probability 1/3.

**Step 2 — Resolve each against the grid geometry** 

At corner cell 0 (top-left):

| Slip direction (prob 1/3 each) | Attempted move | Result |
|---|---|---|
| LEFT (intended) | off the left wall | stays at $0$ |
| UP (orthogonal) | off the top wall | stays at $0$ |
| DOWN (orthogonal) | to the cell below | goes to $4$ |

**Step 3 — Aggregate identical outcomes** 

Two of the three directions keep the agent at 0, while one moves it to 4. Therefore, the transition distribution is:

$\displaystyle p(0\mid 0,\text{LEFT}) = 1/3 + 1/3 = 2/3$    
$\displaystyle p(4\mid 0,\text{LEFT}) = 1/3$        

**Step 4 — Check normalisation**

As a probability distribution over possible next states, the transition probabilities must sum to one:

$\displaystyle \sum_{s’ \in \mathcal{S}} p(s’\mid 0,\text{LEFT}) = \frac{2}{3}+\frac{1}{3}=1$

This is consistent with the transition probabilities provided by the environment:

$\displaystyle P[0][\text{LEFT}] = [(2/3, 0), (1/3, 4)]$

Thus, from state 0, taking LEFT leads to state 0 with probability 2/3 and to state 4 with probability 1/3, accounting for all possible outcomes.

**Key concept**

The transition function is a probability distribution over next states; slip dynamics plus wall geometry determine it, and it must always sum to 1 over the possible next states s'. Aggregating outcomes that land in the same state is the crucial step.

### Exercise 2.2 — Return and the effect of discounting

In an episodic task, suppose that, starting from t=0, the agent experiences the following sequence of rewards:

$\displaystyle R_1=0,\quad R_2=0,\quad R_3=0,\quad R_4=0,\quad R_5=1$

The agent receives no reward during the intermediate steps and a reward of 1 only when it reaches the goal at the final step.

1. Compute the return from the initial state for each of the following discount factors: $\gamma=1,\, \gamma=0.9,\, \gamma=0.5$

2. Then, compare the three results and explain how the discount factor affects the contribution of a delayed reward to the return.

**Step 1 — Write the return**

The episode terminates at t=5, so the discounted return from the initial time step t=0 is:

$\displaystyle G_0=\sum_{k=0}^{4}\gamma^kR_{k+1}$

Since all intermediate rewards are zero and only the final reward is non-zero, the return reduces to:

$\displaystyle G_0=\gamma^4 \, R_5=\gamma^4$

**Step 2 — Evaluate for different discount factors**

Substituting each value of gamma into the return formula gives:

| $\gamma$ | $G_0=\gamma^4$ |
|---|---|
| $1.0$ | $1.0$ |
| $0.9$ | $0.9^4 = 0.6561$ |
| $0.5$ | $0.5^4 = 0.0625$ |

**Step 3 — Interpret**

With gamma=1, a reward has the same weight regardless of when it is received, so the final reward contributes its full value to the return. As gamma decreases, the same delayed reward contributes progressively less to the return. Thus, discounting makes the agent value sooner rewards more than later ones. In addition, for bounded rewards and gamma, discounting ensures that returns remain finite even for infinite-horizon tasks.

**Key concept**

The return is the discounted sum of future rewards. The discount factor gamma determines how much the agent values future rewards relative to immediate ones: the smaller gamma, the more short-sighted the evaluation. For bounded rewards, any gamma<1 also guarantees that the infinite-horizon return converges to a finite value.

**Exercise 2.3 — Infinite-Horizon Return with Constant Rewards**

Consider a continuing task in which the agent receives the same reward at every time step:

$\displaystyle R_{t+1}=r$

The interaction continues indefinitely, and the discount factor satisfies: 

$\displaystyle \gamma\in[0,1)$.

1. Write the infinite-horizon discounted return and derive its closed-form expression.
2. Verify your result using the recursive relation.
3. Evaluate the return for r=1 and gamma=0.9.
4. Explain why the return remains finite even though the agent receives rewards forever.

**Step 1 — Recognize the geometric series**

Since the agent receives the same reward r at every time step, the return is an infinite discounted sum:

$\displaystyle G_t=r+\gamma r+\gamma^2r+\gamma^3r+\cdots$

Each reward received one additional step into the future is multiplied by one additional factor of gamma. Factoring out r gives:

$\displaystyle G_t=r\left(1+\gamma+\gamma^2+\gamma^3+\cdots\right)=r\sum_{k=0}^{\infty}\gamma^k$

The term inside parentheses is a geometric series with common ratio gamma. For gamma<1, its terms become progressively smaller and the infinite series converges:

$\displaystyle \sum_{k=0}^{\infty}\gamma^k=\frac{1}{1-\gamma}$

Therefore,

$\displaystyle G_t=\frac{r}{1-\gamma}$

This shows explicitly why discounting can make an infinite-horizon return finite: although the agent receives rewards forever, the contribution of rewards increasingly far into the future becomes progressively smaller.

**Step 2 — Verify using the recursive relation**

The return can also be written recursively:

$\displaystyle G_t=R_{t+1}+\gamma G_{t+1}$

In this particular task, the agent receives the same reward forever. Therefore, viewed from t or from t+1, the future reward sequence has exactly the same structure:

$\displaystyle r,r,r,r,\ldots$

Consequently, the return is the same at every time step, so we can write:

$\displaystyle G_t = G_{t+1} = G$

Substituting into the recursive relation gives:

$\displaystyle G = r + \gamma G$

Solving for $G$:

$\displaystyle G-\gamma G=r;\Rightarrow;G(1-\gamma)=r;\Rightarrow;G=\frac{r}{1-\gamma}$

We therefore obtain exactly the same result as with the geometric-series derivation. The recursive equation has an intuitive interpretation: the value of the entire future reward stream equals the immediate reward plus the discounted value of everything that follows.

**Step 3 — Evaluate**

For r=1 and gamma=0.9:

$\displaystyle G=1+0.9+0.9^2+0.9^3+\cdots$

Using the closed-form expression:

$\displaystyle G=\frac{1}{1-0.9}=\frac{1}{0.1}=10$

Thus, receiving a reward of 1 forever has a return of 10 when evaluated with gamma=0.9. The agent still receives infinitely many rewards, but rewards far into the future contribute increasingly little to the return.

For comparison, if gamma=1, there would be no discounting:

$\displaystyle G=1+1+1+1+\cdots$

and the return would diverge. This illustrates why gamma<1 is particularly useful for continuing tasks.

**Key concept**

For bounded rewards and gamma<1, discounting makes an infinite-horizon return finite because the contribution of distant rewards decreases geometrically. The identity

$\displaystyle G_t = R_{t+1} + \gamma G_{t+1}$

also reveals an important recursive structure: the return from the present can be expressed using the return from the next time step. This idea (using an estimate of a future quantity to help determine a current one) is the foundation of bootstrapping and will reappear in Bellman equations and temporal-difference learning.

### Exercise 2.4 — From Joint Dynamics to Marginals and Expected Reward

For a fixed state–action pair (s,a), suppose the environment dynamics are described by the following joint distribution:

| next state $s'$ | reward $r$ | $p(s',r\mid s,a)$ |
|---|---|---|
| $s'_1$ | $+5$ | $0.6$ |
| $s'_2$ | $-1$ | $0.3$ |
| $s'_3$ | $\ \ 0$ | $0.1$ |

In this example, each possible next state is associated with a single deterministic reward. Using the joint dynamics, compute:

1. the marginal transition probabilities by summing over all possible rewards;
2. the marginal reward probabilities by summing over all possible next states;
3. the expected immediate reward $r(s,a).

**Step 1 — Marginalize over rewards**

The transition probability tells us the probability of reaching a particular next state, regardless of which reward is received. We obtain it from the joint dynamics by summing over all possible rewards:

$\displaystyle p(s’\mid s,a)=\sum_{r\in\mathcal{R}}p(s’,r\mid s,a)$

For each next state the reward is only one:

$\displaystyle p(s’_1\mid s,a)=p(s’_1,+5\mid s,a)=0.6$

$\displaystyle p(s’_2\mid s,a)=p(s’_2,-1\mid s,a)=0.3$

$\displaystyle p(s’_3\mid s,a)=p(s’_3,0\mid s,a)=0.1$

Thus, the marginal transition distribution is:

$\displaystyle p(s’_1\mid s,a)=0.6,\qquad p(s’_2\mid s,a)=0.3,\qquad p(s’_3\mid s,a)=0.1$

As expected, the probabilities of all possible next states sum to one:

$\displaystyle \sum_{s’\in\mathcal{S}}p(s’\mid s,a)=0.6+0.3+0.1=1$

**Step 2 — Marginalize over next states**

The reward probability tells us the probability of receiving a particular reward, regardless of which next state is reached. We obtain it from the joint dynamics by summing over all possible next states:

$\displaystyle p(r\mid s,a)=\sum_{s’\in\mathcal{S}}p(s’,r\mid s,a)$

For the each reward, we have just one possible next state, so:

$\displaystyle p(+5\mid s,a)=p(s’_1,+5\mid s,a)=0.6$

$\displaystyle p(-1\mid s,a)=p(s’_2,-1\mid s,a)=0.3$

$\displaystyle p(0\mid s,a)=p(s’_3,0\mid s,a)=0.1$

Thus, the marginal reward distribution is:

$\displaystyle p(+5\mid s,a)=0.6,\qquad p(-1\mid s,a)=0.3,\qquad p(0\mid s,a)=0.1$

Again, the probabilities sum to one:

$\displaystyle \sum_{r\in\mathcal{R}}p(r\mid s,a)=0.6+0.3+0.1=1$

**Step 3 — Compute the expected reward**

The expected immediate reward is the probability-weighted average of all possible rewards that can follow the state–action pair. Using the marginal reward distribution obtained in Step 2:

$\displaystyle r(s,a) = \sum_{r\in\mathcal{R}}r,p(r\mid s,a)$

Substituting the possible rewards and their probabilities:

$\displaystyle r(s,a)= (+5)(0.6) + (-1)(0.3) + (0)(0.1) = 3-0.3 = 2.7$

Therefore,

$\displaystyle r(s,a)=2.7$

Notice that this value does not need to be one of the rewards that the agent can actually receive, it represents the average immediate reward we would expect over many repetitions of the same state–action pair.

**Key concept**

The joint dynamics completely characterize the environment’s response to a state–action pair. From this single distribution, we can derive the transition function by marginalizing over rewards, the reward distribution by marginalizing over next states, and the expected immediate reward by taking the expectation of that reward distribution.

### Exercise 2.5 — Probability of a Trajectory under a Policy

Consider the following trajectory:

$\displaystyle \tau=(s_0 ,a_0 ,s_1 ,a_1 ,s_2)$

The agent follows a stochastic policy with:

$\displaystyle \pi(a_0\mid s_0)=0.7, \qquad \pi(a_1\mid s_1)=0.5$

and the environment dynamics are:

$\displaystyle p(s_1\mid s_0,a_0)=0.8, \qquad p(s_2\mid s_1,a_1)=1.0$

1. Compute the probability that this specific trajectory occurs when the agent follows the policy.
2. Identify which factors in this probability are determined by the agent’s policy and which are determined by the environment dynamics.
3. Explain why following a fixed policy does not, in general, determine a unique trajectory.

**Step 1 — Compute the trajectory probability**

For the trajectory 

$\displaystyle \tau=(s_0,a_0,s_1,a_1,s_2)$

the probability of observing exactly this sequence is obtained by multiplying the policy probabilities and the transition probabilities along the trajectory:

$\displaystyle p(\tau\mid\pi)=\pi(a_0\mid s_0) \, p(s_1\mid s_0,a_0) \, \pi(a_1\mid s_1) \, p(s_2\mid s_1,a_1)$

Substituting the given values:

$\displaystyle p(\tau\mid\pi) = 0.7 \cdot 0.8 \cdot 0.5 \cdot 1.0 = 0.28$

Therefore,

$\displaystyle p(\tau\mid\pi)=0.28$

so this specific trajectory occurs with probability 28%.

**Step 2 — Separate policy and environment factors**

The factors:

$\displaystyle \pi(a_0\mid s_0)=0.7,\qquad \pi(a_1\mid s_1)=0.5$

are determined by the agent’s policy: they specify how likely the agent is to choose each action in the corresponding state.

The factors

$\displaystyle p(s_1\mid s_0,a_0)=0.8,\qquad p(s_2\mid s_1,a_1)=1.0$

are determined by the environment dynamics: they specify how likely the environment is to produce each next state after the selected action.

**Step 3 — Interpret**

Following a fixed policy does not generally determine a unique trajectory because randomness can arise from two sources. The policy may be stochastic, so the agent may choose different actions when visiting the same state, and the environment may be stochastic, so the same state–action pair may lead to different next states. Thus, a policy determines a probability distribution over possible trajectories, rather than a single trajectory.

**Key concept**

A trajectory is generated jointly by the agent’s policy and the environment dynamics. The policy determines how actions are selected, while the environment determines the consequences of those actions.
